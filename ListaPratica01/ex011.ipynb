### Exercício 11: Calcule preço médio, máximo e mínimo.

```python
import pandas as pd
```

```python
dataframe = pd.read_csv("dados/produtos.csv")
dataframe
```

```python
preco_medio = dataframe["Preco"].mean()
preco_medio
```

```python
maior_preco = dataframe["Preco"].max()
maior_preco
```

```python
menor_preco = dataframe["Preco"].min()
menor_preco
```

```python
print(f'Preco medio R${preco_medio:.2f}, maior e menor: R${maior_preco} R${menor_preco}')
```
