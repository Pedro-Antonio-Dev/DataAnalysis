### Exercício 17: Crie uma tabela de Categoria x Valor Total em Estoque.

```python
import pandas as pd
```

```python
dataframe = pd.read_csv("dados/produtos.csv")
dataframe
```

```python
dataframe["Valor Total em Estoque"] = dataframe["Preco"] * dataframe["Quantidade"]
dataframe
```

```python
dataframe_estoque = dataframe.groupby("Categoria")["Valor Total em Estoque"].sum().reset_index()
dataframe_estoque
```
