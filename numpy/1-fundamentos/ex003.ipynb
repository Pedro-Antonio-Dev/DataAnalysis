```python
import numpy as np
```

```python
array = np.array([[10, 20, 30], 
                  [40, 50, 60],
                  [70, 80, 90]])

print(array.ndim) # número de dimensões
print(array.shape) # formato do array
array
```

```python
# Buscar elemento na linha 1, coluna 2
elemento = array[1, 2]
print(elemento)
```

```python
# Pegar dados da primeira linha
linha_0 = array[0] # pega a linha 0
print(linha_0)
```

```python
# Pegar dados da segunda coluna
coluna_1 = array[:, 1] # pega a coluna 1
print(coluna_1)
```

```python
# Fatiando para obter um subarray
subarray = array[0:2, 1:3] # pega as linhas 0 e 1, colunas 1 e 2
subarray
```

```python
mascara = array > 50
mascara
```

```python
# Filtragem booleano
elementos_maiores_que_50 = array[mascara]
print(elementos_maiores_que_50)
```

```python
# Modificando elemento na linha 1, coluna 1
array[1, 1] = 100
array
```
