```python
import numpy as np
```

```python
class Array:

    def __init__(self, array: list):
        self.array = np.array(array)

    def show(self):
        print(self.array)

    def findPosition(self, valor):
        return np.where(self.array==valor)
    
    def approveds(self, mean):
        approved = np.where(self.array > mean)
        return self.array[approved]
    
    def maxGrade(self):
        return np.max(self.array)
    
    def minGrade(self):
        return np.min(self.array)
```

```python
array1 = Array([10, 20, 30, 40, 50])
array1.show()
```

```python
# Pesquisando no array
print(array1.findPosition(30))
```

```python
# Exemplo mais real
notas = Array([55, 67, 72, 48, 90, 85, 60, 50])
notas.show()
```

```python
# Pesquisar quais alunos foram aprovados
print(notas.approveds(60))
```

```python
# Notas maxima e minima
print(notas.maxGrade())
print(notas.minGrade())
```
