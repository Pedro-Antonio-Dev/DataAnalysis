Copy -> Uma copia do array é criada em memoria, mas as 2 matrizes são independentes (Original e Copia)

View -> Não é criada uma copia, ela compartilha os mesmos dados do array original

```python
import numpy as np
```

```python
array_original = np.array([[1, 2, 3],
                           [4, 5, 6]])

array_original
```

```python
# Copy
array_copia = array_original.copy()
array_copia
```

```python
array_copia[0, 0] = 100
array_original
```

```python
# View
array_view = array_original[0:, 0:]
array_view[1, 1] = 99
array_original
```
