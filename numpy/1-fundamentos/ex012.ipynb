```python
import numpy as np
```

```python
class Array:

    def __init__(self, array: list):
        self.array = np.array(array)

    def dataTable(self, array1: "Array", array2: "Array", array3: "Array"):
        return np.column_stack((self.array, array1.array, array2.array, array3.array))

    @classmethod
    def means(cls, array1: "Array", array2: "Array"):
        return cls(np.mean(np.column_stack((array1.array, array2.array)), axis=1))
    
    def approveds(self, minMean):
        return self.array[np.where(self.array > minMean)]
```

```python
names = Array(['Pedro', 'Antonio', 'Daniel', 'Monalysa', 'Ana'])
mathGrades = Array([55, 67, 72, 48, 90])
fisisGrades = Array([88, 76, 95, 64, 70])
```

```python
mean = Array.means(mathGrades, fisisGrades)
print(mean.array)
```

```python
approveds = mean.approveds(60)
print(approveds)
```

```python
dt = names.dataTable(mathGrades, fisisGrades, mean)
print(dt)
```
