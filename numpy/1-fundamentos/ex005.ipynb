```python
import numpy as np
```

```python
array = np.array([1, 2, 3, 4, 5])
array
```

```python
# Somar valores dentro do array
soma = np.add(array, 10)
soma
```

```python
# Subtrair valores dentro do array
subtracao = np.subtract(array, 1)
subtracao
```

```python
# Multiplicar valores dentro do array
produto = np.multiply(array, 2)
produto
```

```python
# Dividir valores dentro do array
produto = np.divide(array, 10)
produto
```

```python
# Funções trigonometricas
seno = np.sin(array)
print(seno)

cosseno = np.cos(array)
print(cosseno)

tangente = np.tan(array)
print(tangente)
```

```python
# Funções estatisticas
media = np.mean(array)
print(media)

desvio_padrao = np.std(array)
print(desvio_padrao)
```

```python
# Funções de comparação
maior_3 = array > 3
maior_3
```

```python
# Funções exponenciais
raiz = np.sqrt(array)
print(raiz)

potencia = np.power(array, 2)
print(potencia)

exponencial = np.exp(array)
```

```python
# Funções logaritmicas
logaritmo = np.log(array)
logaritmo
```

```python
# Arrendondamento
arred = np.round(array / 2)
arred
```
