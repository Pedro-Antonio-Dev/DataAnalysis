```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
# Conexao com o DB
engine = create_engine("sqlite:///ecommerce.db")
```

```python
# 1-> Total de vendas por categoria
consulta1 = "SELECT Categoria, SUM(Quantidade) AS Total_Qtd_Vendida FROM produtos GROUP BY Categoria"
resultado1 = pd.read_sql_query(consulta1, engine)
resultado1
```

```python
# 2-> Produto com mais vendas
consulta2 = "SELECT Produto, SUM(Quantidade) AS Total_Qtd_Vendida FROM produtos GROUP BY Produto ORDER BY Total_Qtd_Vendida DESC LIMIT 1"
resultado2 = pd.read_sql_query(consulta2, engine)
resultado2
```

```python
# 3 -> Total de vendas por produto
consulta3 = "SELECT Produto, SUM(Quantidade) AS Total_Qtd_Vendidos FROM produtos GROUP BY Produto ORDER BY Total_Qtd_Vendidos DESC"
resultado3 = pd.read_sql_query(consulta3, engine)
resultado3
```

```python
# 4-> Media de preco por categoria
consulta4 = "SELECT Categoria, AVG(Preço) AS Media_Preco_Categoria FROM produtos GROUP BY Categoria"
resultado4 = pd.read_sql_query(consulta4, engine)
resultado4
```
