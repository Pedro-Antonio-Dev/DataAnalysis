```python
import pandas as pd
```

```python
dataframe = pd.read_json("../arquivosCSV/receitas.json", encoding="utf8")
dataframe
```

```python
# Analise -> Total de ingredientes por receita
receitas_expandidas = []
for receita in dataframe['receitas']:
    total_ingredientes_por_receita = len(receita['ingredientes'])
    receita_informacoes = {
        'titulo': receita['titulo'],
        'tempo_preparo': receita['tempo_preparo'],
        'porcoes': receita['porcoes'],
        'total_ingredientes': total_ingredientes_por_receita
    }
    receitas_expandidas.append(receita_informacoes)
```

```python
receitas_expandidas
```

```python
dataframe_expandido = pd.DataFrame(receitas_expandidas)
dataframe_expandido
```
