```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Nome': ['Pedro', 'Antonio', 'Monalysa', 'Daniel', 'Aninha'],
    'Idade': [19, 29, 19, 13, 47],
    'Cargo': ['CEO', 'Gerente', 'Analista', 'Coordenador', 'Analista'],
    'Salario': [5000, 8000, 15000, 4800, 6000]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.describe()
```
