```python
dados_notas = {
    'Nome': ['Pedro', 'Antonio', 'Monalysa'],
    'Matematica': [85, 90, 92],
    'Fisica': [91, 89, 97],
    'Quimica': [78, 81, 87]
}
```

```python
df_notas = pd.DataFrame(dados_notas)
df_notas
```

```python
# Calculando a media das notas
df_notas['Media'] = df_notas[['Matematica', 'Fisica', 'Quimica']].mean(axis=1)
df_notas
```

```python
# Alunos com media maior que 90
alunos_notas_boas = df_notas[df_notas['Media'] >= 90]
alunos_notas_boas
```
