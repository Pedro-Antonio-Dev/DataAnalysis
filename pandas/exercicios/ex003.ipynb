```python
import pandas as pd
```

```python
dados = pd.read_json("../datasets/arquivosCSV/dadosProdutos.json", encoding="utf8")
dados
```

```python
# Filtro -> Produto mais caro
produto_mais_caro = max(dados['produtos'], key=lambda produto: produto['preco'])
produto_mais_caro
```

```python
# Normalizando meus dados
dataframe = pd.json_normalize(dados['produtos'])
dataframe
```

```python
# Filtro -> Produto mais bem avaliado
produto_melhor_avaliado = dataframe.loc[dataframe['avaliacao'].idxmax()].reset_index()
produto_melhor_avaliado
```

```python
# Somando -> quantidade total em estoque
quantidade_estoque = sum(dataframe['estoque'])
quantidade_estoque
```
