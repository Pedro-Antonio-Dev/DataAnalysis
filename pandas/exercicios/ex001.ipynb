```python
import pandas as pd
```

```python
df_vendas = pd.DataFrame({
    'Produto': ['Notebook', 'Mouse', 'Notebook', 'Teclado', 'Mouse', 'Teclado'],
    'Quantidade': [2, 5, 1, 3, 2, 4],
    'Receita': [6000, 500, 3000, 900, 200, 1200],
    'Meses': ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']
})

df_vendas
```

```python
df_vendas_tratado = df_vendas.groupby('Produto')[['Quantidade', 'Receita']].sum().reset_index()
df_vendas_tratado
```

```python
df_vendas_media = df_vendas.groupby('Produto')['Quantidade'].mean().reset_index()
df_vendas_media
```

```python
produto_mais_vendido = df_vendas.groupby('Produto')['Quantidade'].sum().idxmax()
produto_mais_vendido
```

```python
maior_quantidade_vendida = df_vendas.groupby('Produto')['Receita'].sum().idxmax()
maior_quantidade_vendida
```
