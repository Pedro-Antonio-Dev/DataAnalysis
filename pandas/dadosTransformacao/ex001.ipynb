```python
import pandas as pd
```

```python
data_frame = pd.read_json("dados/dados_hospedagem.json")
```

```python
data_frame.head()
```

```python
data_frame = pd.json_normalize(data_frame["info_moveis"])
data_frame
```

```python
collumns = list(data_frame.columns)
collumns
```

```python
data_frame = data_frame.explode(collumns[3:])
data_frame
```

```python
data_frame.reset_index(inplace=True, drop=True)
```

```python
data_frame.head()
```

```python
data_frame.dtypes
```

### Conversão dos Dados Numéricos

```python
import numpy as np
```

```python
data_frame["max_hospedes"] = data_frame["max_hospedes"].astype(np.int64)
```

```python
collumns_numerics = ["quantidade_banheiros", "quantidade_quartos", "quantidade_camas"]
```

```python
data_frame[collumns_numerics] = data_frame[collumns_numerics].astype(np.int64)
```

```python
data_frame.dtypes
```

```python
data_frame["avaliacao_geral"] = data_frame["avaliacao_geral"].astype(np.float64)
```

```python
data_frame.dtypes
```

### Convertendo os preços ($0.0) em Float

```python
data_frame["preco"] = data_frame["preco"].apply(lambda x: x.replace("$", "").replace(",", "").strip())
```

```python
data_frame["preco"]
```

```python
data_frame["preco"] = data_frame["preco"].astype(np.float64)
data_frame.dtypes
```

```python
collumns_taxas = ["taxa_deposito", "taxa_limpeza"]
```

```python
for col in collumns_taxas:
    data_frame[col] = (
        data_frame[col]
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
    )
```

```python
data_frame[collumns_taxas] = data_frame[collumns_taxas].astype(np.float64)
```

```python
data_frame.dtypes
```

### Conversão dos Dados Textuais

```python
data_frame["descricao_local"]
```

```python
data_frame["descricao_local"] = data_frame["descricao_local"].str.lower()
data_frame["descricao_local"]
```

```python
data_frame["descricao_local"] = data_frame["descricao_local"].str.replace("[^a-zA-Z0-9\-\']", " ", regex=True)
```

```python
data_frame["descricao_local"][3000]
```

```python
data_frame["descricao_local"] = data_frame["descricao_local"].str.replace("(?<!\w)-(?!\w)", " ", regex=True)
```

```python
data_frame["descricao_local"]
```

```python
data_frame["descricao_local"] = data_frame["descricao_local"].str.split()
```

```python
data_frame
```

```python
data_frame["comodidades"] = data_frame["comodidades"].str.replace("\{|}|\'", "", regex=True)
```

```python
data_frame["comodidades"] = data_frame["comodidades"].str.split(",")
```

```python
data_frame["comodidades"]
```
