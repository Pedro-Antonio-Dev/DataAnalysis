```python
# Exemplo real
vendas_mensais = pd.Series(
    [2000, 2500, 3000, 2800, 3500, 5000],
    index=['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun']
)
```

```python
vendas_mensais
```

```python
# Media das vendas
media_vendas = np.mean(vendas_mensais)
media_vendas
```

```python
# Mes com mais vendas
mesMaiorVendas = vendas_mensais.idxmax()
mesMaiorVendas
```

```python
vendas_acima_media = vendas_mensais[vendas_mensais > media_vendas]
vendas_acima_media
```
