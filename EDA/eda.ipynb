### Transformação dos Dados

```python
df_contratos.dtypes
```

```python
# Força a conversão dos dados, transformando os erros em valores nulos 
df_contratos["TotalCharges"] = pd.to_numeric(df_contratos["TotalCharges"], errors="coerce")
df_contratos.dtypes
```

```python
df_contratos.info()
```

### Renomear Colunas

```python
df_clientes.info()
```

```python
df_clientes = df_clientes.rename(columns={"SeniorCitizen": "ElderlyPerson"})
df_clientes
```

```python
df_clientes.rename(columns={"Partner": "Married"}, inplace=True)
df_clientes
```

```python
df_clientes.columns = ["IDCliente", "Genero", "PessoaIdosa", "Casado(a)", "Dependentes"]
df_clientes
```

```python
df_clientes.info()
```

### Juntando DataFrames

```python
df_servicos.head()
```

```python
df_servicos.rename(columns={"customerID": "IDCliente"}, inplace=True)
```

```python
df_servicos.info()
```

```python
df_temp = df_clientes.merge(df_servicos, on=["IDCliente"])
df_temp.info()
```

```python
df_churn_temp = df_temp.merge(df_contratos, left_on=["IDCliente"], right_on=["customerID"])
df_churn_temp.info()
```

```python
df_churn = df_clientes.merge(df_servicos, on=["IDCliente"]).merge(df_contratos, left_on=["IDCliente"], right_on=["customerID"])
df_churn
```

```python
df_churn.drop(["customerID"], inplace=True, axis=1) # axis=1 -> opera nas colunas
df_churn.info()
```

### Tratando Valores Ausentes

```python
df_churn.isna().sum()
```

```python
# Pega os registros que possui valores NaN
df_churn[df_churn.isna().any(axis=1)]
```

```python
# Remover coluna com valor ausente
df_churn.dropna(axis=1)
```

```python
# Remover coluna se todos os valores forem NaN
df_churn.dropna(axis=1, how="all")
```

```python
# Remover linhas com valores NaN
df_churn.dropna(axis=0)
```

```python
# Remover linhas com todos os valores ausentes
df_churn.dropna(axis=0, how="all")
```

```python
# Trocando os valores por 0
df_churn.fillna(0)
```

```python
# Especifica a coluna que vai ser modificada
df_churn.fillna(value={"TotalCharges": 0})
```

```python
# Preenchendo pela media
df_churn.fillna(value={"TotalCharges": df_churn["TotalCharges"].mean()})
```

### Filtros e Agrupamentos de Valores

```python
# Quais clientes tem apenas 1 mes de contrato
len(df_churn[df_churn["tenure"] == 1])
```

```python
# Representacao da informacao em porcentagem
len(df_churn[df_churn["tenure"] == 1]) / len(df_churn) * 100
```

```python
# Clientes que possuem entre 1 a 5 meses de contrato
len(df_churn[(df_churn["tenure"] >= 1) & (df_churn["tenure"] <= 5)])
```

```python
# Quantidade de clientes por tempo de contrato
df_churn.groupby(["tenure"])["tenure"].count().sort_values(ascending=False)
```
