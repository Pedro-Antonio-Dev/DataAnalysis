```python
import pandas as pd
```

### Definindo os Dados

```python
dict_metrics = {
    "idade": [15, 18, 25, 26, 40, 40, 58, 60],
    "altura": [160, 165, 149, 168, 172, 174, 174, 178]
}
```

```python
type(dict_metrics)
```

```python
dataframe = pd.DataFrame(dict_metrics)
dataframe
```

### Medidas de Posicao

- Media: Soma de todos os valores dividida pelo numero de valores. Pode ser sensivel a valores extremos
- Mediana: E o valor do meio quando os dados estao ordenados. Nao sofre por valores extremos
- Moda: Valor com a maior ocorrencia

```python
media_idade = dataframe["idade"].mean()
media_altura = dataframe["altura"].mean()
print(f"Media de idades: {media_idade:.1f}")
print(f"Media de alturas: {media_altura:.1f}")
```

```python
mediana_idade = dataframe["idade"].median()
mediana_altura = dataframe["altura"].median()
print(f"Mediana de idades: {mediana_idade:.1f}")
print(f"Mediana de alturas: {mediana_altura:.1f}")
```

```python
moda_idade = dataframe["idade"].mode()
moda_altura = dataframe["altura"].mode()
print(f"Moda de idades: {moda_idade}")
print(f"Moda de alturas: {moda_altura}")
```

### Medidas de Dispersão
- Variância: É a media dos quadrados entre cada valor e a media. Informa o quão distantes os valores estão da media.
- Desvio Padrão: É a raiz quadrada da Variância. Expressa a dispersão em termos da mesma unidade dos dados.
- Coeficiente de Variação: É o Desvio Padrão dividido pela media, expresso em porcentagem. Indica a variabilidade relativa dos dados em relação a media

```python
variancia_idade = dataframe["idade"].var()
print(f"Variancia de idade: {variancia_idade}")

variancia_altura = dataframe["altura"].var()
print(f"Variancia de altura: {variancia_altura}")
```

```python
desvio_padrao_idade = dataframe["idade"].std()
print(f"Desvio Padrao das idades: {desvio_padrao_idade}")

desvio_padrao_altura = dataframe["altura"].std()
print(f"Desvio Padrao das alturas: {desvio_padrao_altura}")
```

```python
coeficiente_variacao_idade = desvio_padrao_idade / media_idade * 100
print(f"coeficiente de variacao das idades: {coeficiente_variacao_idade:.2f}%")

coeficiente_variacao_altura = desvio_padrao_altura / media_altura * 100
print(f"coeficiente de variacao das alturas: {coeficiente_variacao_altura:.2f}%")
```

### Medidas de Forma (Distribuição dos Dados)
- Assimetria: Indica o grau e direção da distorção da distribuição em relação à media. Se positiva, indica a cauda direita da distribuição mais longe, a oposta indica a cauda esquerda mais longa
- Curtose: Mede o pico ou "pontuação" da distribuição. Uma curtose alta indica uma distribuição mais concentreada, já o oposto indica uma distribuição mais achatada

```python
dataframe
```

```python
assimetria_idade = dataframe["idade"].skew()
print(f"Assimetria das idades: {assimetria_idade}")

assimetria_altura = dataframe["altura"].skew()
print(f"Assimetria das alturas: {assimetria_altura}")
```

##### Assimetria das idades positiva pois o valores maiores estao mais espalhados
##### Assimetria das alturas negativa pois o valores maiores estao mais proximos

```python
curtose_idade = dataframe["idade"].kurtosis()
print(f"Curtose das idades: {curtose_idade}")

curtose_altura = dataframe["altura"].kurtosis()
print(f"Curtose das alturas: {curtose_altura}")
```

##### Curtose das idades negativa pois o dados sao mais espalhados
##### Curtose das alturas positiva pois o dados estao mais proximos da media

### Correlação
- Mede a relação entre duas variaveis, se positiva = diretamente proporcionais. Se negativa = inversamente proporcionais

### Índices de Correlação
- Coeficiente de Pearson: Mede a relação linear entre duas variaveis, variando de -1 a 1 e 0 para nenhuma correlação
- Coeficiente de Spearman: Util quando os dados nao tem uma relação linear clara ou quando as variaveis não são numericamente escalonaveis
- Pearson para relação liner e variaveis continuas e Spearman para relação não linear, ordinais ou com outliers

```python
dataframe.corr(method="pearson")
```

```python
dataframe.corr(method="spearman")
```

```python
dataframe["idade"].corr(dataframe["altura"], method="pearson")
```

```python
dataframe["idade"].corr(dataframe["altura"], method="spearman")
```
