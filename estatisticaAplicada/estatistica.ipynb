```python
import pandas as pd
```

### Definindo os Dados

```python
dict_metrics = {
    "idade": [15, 18, 25, 26, 40, 40, 58, 60],
    "altura": [160, 165, 149, 168, 172, 174, 174, 178]
}
```

```python
type(dict_metrics)
```

```python
dataframe = pd.DataFrame(dict_metrics)
dataframe
```

### Medidas de Posicao

- Media: Soma de todos os valores dividida pelo numero de valores. Pode ser sensivel a valores extremos
- Mediana: E o valor do meio quando os dados estao ordenados. Nao sofre por valores extremos
- Moda: Valor com a maior ocorrencia

```python
media_idade = dataframe["idade"].mean()
media_altura = dataframe["altura"].mean()
print(f"Media de idades: {media_idade:.1f}")
print(f"Media de alturas: {media_altura:.1f}")
```

```python
mediana_idade = dataframe["idade"].median()
mediana_altura = dataframe["altura"].median()
print(f"Mediana de idades: {mediana_idade:.1f}")
print(f"Mediana de alturas: {mediana_altura:.1f}")
```

```python
moda_idade = dataframe["idade"].mode()
moda_altura = dataframe["altura"].mode()
print(f"Moda de idades: {moda_idade}")
print(f"Moda de alturas: {moda_altura}")
```
